* `Score` should contain the count of correct answers you have obtained in a particular reading session.

```python
# @title IELTS Reading Progress examiner

import matplotlib.pyplot as pl,numpy as np

score=[35,38,38,37,40,37,37,38,38] # @param {type:"raw", placeholder:"Enter score as [1,2,3..]"}
score=np.array(score)
time =[44,57,42,49,43,47,45,52,54] # @param {type:"raw", placeholder:"Enter time taken as [1,2,3..]"}
time=np.array(time)

# Band score calculator
def bandscore(rawvalue):
  final=4
  if 13<=rawvalue<=14:final+=0.5
  elif 15<=rawvalue<=18:final+=1.0
  elif 19<=rawvalue<=22:final+=1.5
  elif 23<=rawvalue<=26:final+=2.0
  elif 27<=rawvalue<=29:final+=2.5
  elif 30<=rawvalue<=32:final+=3.0
  elif 33<=rawvalue<=34:final+=3.5
  elif 35<=rawvalue<=36:final+=4.0
  elif 37<=rawvalue<=38:final+=4.5
  elif 39<=rawvalue<=40:final+=5
  return final

band = np.array([bandscore(data) for data in score])

# Attempts taken - to be used for x axis.
attempts=np.arange(1,len(score)+1,1)

# Multiple plot axes in same figure
fig,ax1 = pl.subplots(figsize=(8,5))
ax2=ax1.twinx()
ax3=ax1.twinx()

# Tuning limit and labels for different data
ax1.set_ylim(0,40)
ax2.set_ylim(0,60)
ax3.set_ylim(0,9)

ax1.set_xlabel("Attempts")
ax1.set_ylabel("Raw Score")
ax2.set_ylabel("Time Taken")
ax3.set_ylabel("Band Score")

# Colours to distinguish between datas
clr1,clr2,clr3="blue","red","lime"

ax1.yaxis.label.set_color(clr1)
ax2.yaxis.label.set_color(clr2)
ax3.yaxis.label.set_color(clr3)

# Plotting and legends
p1=ax1.plot(attempts,score, marker='.', color=clr1, label="Raw Score")
p2=ax2.plot(attempts,time, marker='.', color=clr2, label="Time Taken")
p3=ax3.plot(attempts, band, marker='.', color=clr3, label="Band Score")

ax3.spines['right'].set_position(('outward',60))

ax1.legend(handles=p1+p2+p3,loc='best')
pl.show()

# Printing a table

print("------------------------------------------------------------------")
print('{:<1}{:<16}{:<1}{:<15}{:<1}{:<15}{:<1}{:<15}{:<1}'.format("|","Attempt","|","Time Taken ","|","Raw Score","|","Band Score","|"))
print("------------------------------------------------------------------")
for att,tt,sc,bs in zip(attempts,time,score,band):
  print('{:<1}{:<16}{:<1}{:<15}{:<1}{:<15}{:<1}{:<15}{:<1}'.format("|",att,"|",tt,"|",sc,"|",bs,"|"))
print("------------------------------------------------------------------")

# Average Values
for val, txt in zip([time, score, band],['time (min.)', 'raw score', 'band score']):
  avg=np.sum(val)/len(val)
  avg=np.round(avg,2)
  print('{:<30}{:<2}{:<5}'.format(f"Average {txt} recorded ","= ",avg))
```
